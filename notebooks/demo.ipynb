# EnhanceX Framework Interactive Demonstration
Welcome to the EnhanceX interactive Jupyter notebook demonstration.

```python
import cv2
import numpy as np
from enhancex import ImageEnhancer, VideoEnhancer, SuperResolutionEngine

# Create sample input image
sample = np.zeros((300, 300, 3), dtype=np.uint8)
cv2.circle(sample, (150, 150), 50, (0, 255, 0), -1)

# Apply Image Enhancement
enhancer = ImageEnhancer()
enhanced = enhancer.enhance(sample, sharpen=1.5, clahe=True)
print('Enhanced Image shape:', enhanced.shape)
```
